# Project: Diabetes Risk & HbA1c Progression Engine


## Phase 1: Data Preprocessing & EDA.

````python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from scipy import stats
````

````python
url = "https://raw.githubusercontent.com/jbrownlee/Datasets/master/pima-indians-diabetes.data.csv"
column_names = [
    'Pregnancies', 'Glucose', 'BloodPressure', 'SkinThickness',
    'Insulin', 'BMI', 'DiabetesPedigreeFunction', 'Age', 'Outcome'
]
df = pd.read_csv(url, names=column_names)

print(f"Initial dataset shape: {df.shape}")
display(df.head())
````

````python
zero_columns = ['Glucose', 'BloodPressure', 'SkinThickness', 'Insulin', 'BMI']
print("\nNumber of invalid zero values per column:")
for col in zero_columns:
    num_zeros = (df[col] == 0).sum()
    print(f" - {col}: {num_zeros}")
````

````python
for col in zero_columns:
    df[col] = df[col].replace(0, np.nan)
    df[col] = df[col].fillna(df[col].median())
````

````python
continuous_cols = ['Pregnancies', 'Glucose', 'BloodPressure', 'SkinThickness', 'Insulin', 'BMI', 'DiabetesPedigreeFunction', 'Age']
z_scores = np.abs(stats.zscore(df[continuous_cols]))
filtered_entries = (z_scores < 3).all(axis=1)
df_cleaned = df[filtered_entries].copy()

print(f"\nShape after Z-score outlier removal (|Z| <= 3): {df_cleaned.shape}")
````

````python
print("\nSummary Statistics of cleaned dataset:")
display(df_cleaned.describe())
````

````python
plt.figure(figsize=(10, 8))
sns.heatmap(df_cleaned.corr(), annot=True, cmap='coolwarm', fmt='.2f', linewidths=0.5)
plt.title('Correlation Heatmap (Cleaned Pima Dataset)')
plt.show()
````

## Phase 2: Feature Engineering & Target Formulation

- **Continuous target**: Estimated $HbA1c$ (%) based on the clinically accepted ADAG (A1C-Derived Average Glucose) formula:
  $$HbA1c = \frac{Glucose + 46.7}{28.7}$$



````python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
````

````python
df_cleaned['HbA1c'] = (df_cleaned['Glucose'] + 46.7) / 28.7
````

````python
X = df_cleaned.drop(columns=['HbA1c', 'Outcome'])
y_reg = df_cleaned['HbA1c']
y_clf = df_cleaned['Outcome']
````

````python
X_train, X_test, y_train_reg, y_test_reg, y_train_clf, y_test_clf = train_test_split(
    X, y_reg, y_clf, test_size=0.2, random_state=42, stratify=y_clf
)
````

````python
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

print(f"Training set shape: {X_train_scaled.shape}")
print(f"Test set shape: {X_test_scaled.shape}")
````

## Phase 3: Regression Modeling (HbA1c Estimation)


````python
from sklearn.linear_model import LinearRegression, Ridge, PoissonRegressor
from sklearn.model_selection import GridSearchCV
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score
````

````python
reg_results = {}
````

````python
def evaluate_regressor(name, y_true, y_pred):
    rmse = np.sqrt(mean_squared_error(y_true, y_pred))
    mae = mean_absolute_error(y_true, y_pred)
    r2 = r2_score(y_true, y_pred)
    reg_results[name] = {'RMSE': rmse, 'MAE': mae, 'R²': r2}
````

````python
mlr = LinearRegression()
mlr.fit(X_train_scaled, y_train_reg)
y_pred_mlr = mlr.predict(X_test_scaled)
evaluate_regressor('Multiple Linear Regression', y_test_reg, y_pred_mlr)
````

````python
ridge = Ridge()
param_grid_ridge = {'alpha': np.logspace(-4, 4, 100)}
ridge_cv = GridSearchCV(ridge, param_grid_ridge, cv=5, scoring='neg_mean_squared_error')
ridge_cv.fit(X_train_scaled, y_train_reg)
best_ridge = ridge_cv.best_estimator_
y_pred_ridge = best_ridge.predict(X_test_scaled)
evaluate_regressor('Ridge Regression (Tuned)', y_test_reg, y_pred_ridge)
````

````python
poisson_glm = PoissonRegressor(max_iter=1000)
poisson_glm.fit(X_train_scaled, y_train_reg)
y_pred_poisson = poisson_glm.predict(X_test_scaled)
evaluate_regressor('Poisson GLM Regressor', y_test_reg, y_pred_poisson)
````

````python
reg_df = pd.DataFrame(reg_results).T
print("=== Regression Model Performance (HbA1c Estimation) ===")
display(reg_df)
print(f"\nBest Ridge alpha found: {ridge_cv.best_params_['alpha']:.4f}")
````

## Phase 4: Classification Modeling & Discriminant Analysis (Diabetes Risk)


````python
from sklearn.linear_model import LogisticRegression
from sklearn.discriminant_analysis import LinearDiscriminantAnalysis, QuadraticDiscriminantAnalysis
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, roc_auc_score
````

````python
clf_results = {}
````

````python
def evaluate_classifier(name, model, X_tst, y_tst):
    y_pred = model.predict(X_tst)
    y_prob = model.predict_proba(X_tst)[:, 1]

    acc = accuracy_score(y_tst, y_pred)
    prec = precision_score(y_tst, y_pred)
    rec = recall_score(y_tst, y_pred)
    f1 = f1_score(y_tst, y_pred)
    auc = roc_auc_score(y_tst, y_prob)

    clf_results[name] = {
        'Accuracy': acc,
        'Precision': prec,
        'Recall': rec,
        'F1-Score': f1,
        'AUC-ROC': auc
    }
````

````python
log_reg = LogisticRegression(solver='liblinear', random_state=42)
param_grid_lr = {'C': np.logspace(-4, 4, 100), 'penalty': ['l1', 'l2']}
log_reg_cv = GridSearchCV(log_reg, param_grid_lr, cv=5, scoring='accuracy')
log_reg_cv.fit(X_train_scaled, y_train_clf)
best_lr = log_reg_cv.best_estimator_
evaluate_classifier('Logistic Regression (Tuned)', best_lr, X_test_scaled, y_test_clf)
````

````python
lda = LinearDiscriminantAnalysis()
lda.fit(X_train_scaled, y_train_clf)
evaluate_classifier('Linear Discriminant Analysis', lda, X_test_scaled, y_test_clf)
````

````python
qda = QuadraticDiscriminantAnalysis()
qda.fit(X_train_scaled, y_train_clf)
evaluate_classifier('Quadratic Discriminant Analysis', qda, X_test_scaled, y_test_clf)
````

````python
clf_df = pd.DataFrame(clf_results).T
print("=== Classification Model Performance (Diabetes Risk) ===")
display(clf_df)
print(f"\nBest Logistic Regression hyperparameters: {log_reg_cv.best_params_}")
````

## Phase 5: Custom Inference Function



````python
def evaluate_patient(pregnancies, glucose, bp, skin, insulin, bmi, dpf, age):
    """
    Evaluates single patient physiological metrics for classification risk.
    """
                                                                      
    raw_data = pd.DataFrame([[pregnancies, glucose, bp, skin, insulin, bmi, dpf, age]],
                            columns=X.columns)
    patient_scaled = scaler.transform(raw_data)

                                                                           
    diabetic_prob = best_lr.predict_proba(patient_scaled)[0][1]
    risk_label = "High Risk / Diabetic" if diabetic_prob >= 0.5 else "Low Risk / Healthy"

                                                     
    print("==================================================")
    print("           DIABETES RISK ENGINE RESULT            ")
    print("==================================================")
    print(f" Clinical Stratification:   {risk_label}")
    print(f" Confidence Probability:    {diabetic_prob * 100:.2f}%")
    print("==================================================")
````

```markdown
## Phase 9: Launching the Diagnostic Portal Locally

Run the cell below to start a lightweight Python web server on port `8000`. This will allow you to test your interactive dashboard directly within Colab using a secure local proxy URL.
```

````python
import http.server
import socketserver
import threading
from google.colab import output

PORT = 8000
max_attempts = 10
httpd = None
````

````python
class ThreadingHTTPServer(socketserver.ThreadingMixIn, http.server.HTTPServer):
    allow_reuse_address = True
````

````python
for attempt in range(max_attempts):
    try:
        handler = http.server.SimpleHTTPRequestHandler
        httpd = ThreadingHTTPServer(("", PORT), handler)
        break
    except OSError:
        print(f"Port {PORT} is already occupied. Trying next port...")
        PORT += 1

if httpd:
    def run_server():
        print(f"Serving local diagnostic portal on port {PORT}...")
        httpd.serve_forever()

                                                                                  
    server_thread = threading.Thread(target=run_server, daemon=True)
    server_thread.start()

                                                                           
    local_url = output.eval_js(f'google.colab.kernel.proxyPort({PORT})')
    print("\n=======================================================")
    print("👉 CLICK THE LINK BELOW TO OPEN YOUR LOCAL DASHBOARD:")
    print(f"{local_url}")
    print("=======================================================")
else:
    print("Could not allocate an available port for the local server.")
````
